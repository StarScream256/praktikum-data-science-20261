```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

### Informasi dataset

```python
df = sns.load_dataset('titanic')
df.head(5)
# df.info()
# df.describe()
```

### Menangani nilai hilang (missing value)

```python
df.isnull().sum()
```

```python
# menghapus kolom deck karena memiliki banyak nilai null
df = df.drop('deck', axis=1)
# mengisi nilai null pada kolom age dengan median
df['age'] = df['age'].fillna(df['age'].median())
# mengisi nilai null pada kolom embarked dengan modus
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
df['embark_town'] = df['embark_town'].fillna(df['embark_town'].mode()[0])
df.isnull().sum()
```

### Menangani data duplikat

```python
print(f'Baris duplikat : {df.duplicated().sum()}')
```

```python
df = df.drop_duplicates()
print(f'Baris duplikat : {df.duplicated().sum()}')
print(f'Jumlah baris setelah menghapus duplikat : {len(df)}')
```

### Validasi dan konversi tipe data

```python
df.info()

# Konversi survived menjadi tipe data boolean
df['survived'] = df['survived'].astype(bool)
df.info()
```

### Sclaing numerik (normalisasi dan standarisasi)

```python
scaler = MinMaxScaler()
df['fare_normalized'] = scaler.fit_transform(df[['fare']])
print("Normalisasi dengan MinMaxScaler:")
df[['fare', 'fare_normalized']].head()
```

```python
scaler = StandardScaler()
df['fare_standardized'] = scaler.fit_transform(df[['fare']])
print("Standardisasi dengan StandardScaler:")
df[['fare', 'fare_standardized']].head()
```

### Menyimpan data bersih

```python
df.to_csv('titanic_cleaned.csv', index=False)
```
