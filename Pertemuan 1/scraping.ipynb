```python
import pandas as pd
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
```

### Membuat URL pencarian

```python
def get_url(page):
    if page == 1:
        return 'https://www.cnnindonesia.com/search/?query=IHSG'
    else:
        return f'https://www.cnnindonesia.com/search/?query=IHSG&page={page}'
```

### Menyimpan data ke csv

```python
def save_to_csv(titles):
    df = pd.DataFrame(titles, columns=['Title'])
    df.to_csv("news_title.csv", index=False, encoding='utf-8-sig')
    print(f"{len(df)} Data saved to news_title.csv")
```

### Scraping judul berita

```python
def scrape_titles():
    options = Options()
    options.add_argument('--disable-gpu')
    driver = webdriver.Chrome(options=options)
    titles = []
    for page in range(1, 10):
        url = get_url(page)
        print(f'Scraping {url}')
        driver.get(url)
        time.sleep(2)
        try:
            h2_elements = driver.find_elements(By.CLASS_NAME,
            'text-cnn_black_light')
            if not h2_elements:
                print(f'Tidak ada judul ditemukan di halaman{page}.')
                break
            for h2 in h2_elements:
                titles.append(h2.text)
        except Exception as e:
            print(f'Error di halaman {page}: {e}')
        continue
    driver.quit()
    return titles
```

```python
results = scrape_titles()
save_to_csv(results)
```
